**Question:** Does applying a log transformation reduce the skewness of the financial features?

```python
financial_features = [
    'Sales',
    'Order Item Discount',
    'Order Item Total',
    'Product Price'
]

skew_comparison = pd.DataFrame({
    'Original Skewness': df[financial_features].skew(),
    'Log1p Skewness': df[financial_features].apply(
        lambda x: np.log1p(x)
    ).skew()
})

print(skew_comparison)
```

**Insight:** The `log1p()` transformation substantially reduced the right skewness of all four financial features. `Sales` decreased from **2.884 to -0.679**, `Order Item Discount` from **3.040 to -0.524**, `Order Item Total` from **2.888 to -0.663**, and `Product Price` from **3.191 to 0.333**. Although the transformation slightly over-corrected the skewness of some features, the resulting distributions were considerably less skewed.

**Decision:** The original numerical features will be **retained for the initial feature set** rather than automatically replacing them with log-transformed versions. Log transformation may be evaluated later for models where it could improve performance. The identified extreme values will **not be removed or arbitrarily capped**, as they appear to represent plausible observations rather than obvious data-quality errors.

**Key Takeaway:** Log transformation can effectively reduce skewness, but **reduced skewness does not automatically mean better model performance**. The final decision will be validated during model development.

**Question:** Are `Shipping Mode` and `Days for shipment (scheduled)` redundant?

```python
shipping_mapping = (
    df.groupby('Shipping Mode')['Days for shipment (scheduled)']
      .nunique()
      .sort_values()
)

print("Number of scheduled-day values for each Shipping Mode:")
print(shipping_mapping)

print("\nDetailed mapping:")

display(
    df[['Shipping Mode', 'Days for shipment (scheduled)']]
    .drop_duplicates()
    .sort_values(['Shipping Mode', 'Days for shipment (scheduled)'])
)
```

**Insight:** The analysis confirmed a one-to-one mapping between `Shipping Mode` and `Days for shipment (scheduled)`. Each shipping mode corresponds to exactly one scheduled shipping duration: `Same Day` = 0 days, `First Class` = 1 day, `Second Class` = 2 days, and `Standard Class` = 4 days. Therefore, the two features contain the same underlying information.

**Decision:** `Days for shipment (scheduled)` will be retained as the initial model feature, while `Shipping Mode` will be excluded from the final feature set to avoid duplicate information.



**Question:** Does `Product Card Id` contain the same information as `Category Id` and `Department Id`?

```python
product_category_mapping = (
    df.groupby('Product Card Id')['Category Id']
      .nunique()
)

print("Product Card Id → Category Id")
print("Products with multiple categories:",
      (product_category_mapping > 1).sum())

print("Maximum categories per product:",
      product_category_mapping.max())

product_department_mapping = (
    df.groupby('Product Card Id')['Department Id']
      .nunique()
)

print("Product Card Id → Department Id")
print("Products with multiple departments:",
      (product_department_mapping > 1).sum())

print("Maximum departments per product:",
      product_department_mapping.max())


```

**Insight:** The analysis confirmed that every `Product Card Id` maps to exactly one `Category Id` and exactly one `Department Id`. No product was found to belong to multiple categories or departments. Therefore, `Product Card Id` deterministically identifies its corresponding category and department within this dataset.

However, these features represent different levels of the product hierarchy: `Department Id` represents a broader grouping, `Category Id` represents a more specific grouping, and `Product Card Id` identifies the individual product.

**Decision:** `Department Id`, `Category Id`, and `Product Card Id` will be retained as categorical features in the initial baseline. Their usefulness and potential redundancy can be evaluated further during model development.

**Question:** Do the date-derived features contain useful information for predicting late deliveries, and should any be removed or transformed?

```python
date_analysis_features = [
    'Order Month',
    'Order Day of Week',
    'Order Hour'
]

for col in date_analysis_features:
    print(f"\n{'=' * 50}")
    print(f"Late delivery rate by {col}")
    print('=' * 50)

    result = (
        df.groupby(col)['Late_delivery_risk']
        .agg(['count', 'mean'])
        .reset_index()
    )

    result['Late Delivery Rate (%)'] = result['mean'] * 100

    display(
        result.drop(columns='mean')
        .round(2)
    )

```

**Insight:** The late-delivery rates across months ranged from **54.06% to 55.79%**, while the rates across weekdays ranged from **54.30% to 55.17%**. These relatively small differences suggest weak standalone associations between the target and these calendar features.

However, `Order Hour` showed more noticeable variation. The late-delivery rate was approximately **50.87% at 08:00** and **59.50% at 12:00**. This suggests that order time may contain useful predictive information, although the observed association does not establish causation.

`Order Year` and `Order Day` will also be retained provisionally because their usefulness has not been fully established. The relatively small number of observations in 2018 should be considered when interpreting year-related patterns.

**Decision:** All five date-derived features—`Order Year`, `Order Month`, `Order Day`, `Order Day of Week`, and `Order Hour`—will be retained provisionally for the initial model. Cyclical encoding may be evaluated for month, weekday, and hour during model experimentation. The final selection will depend on validation performance.

**Question:** Should `Customer Id` and `Order Item Id` be included as predictive features?

```python
identifier_features = [
    'Customer Id',
    'Order Item Id'
]

for col in identifier_features:
    print(f"\n{col}")
    print("Unique values:", df[col].nunique())
    print("Total rows:", len(df))
    print("Duplicate values:", df[col].duplicated().sum())
```

**Insight:** `Order Item Id` contains 180,519 unique values across 180,519 rows, making it a unique identifier for each record. It does not directly describe shipment characteristics and could encourage the model to learn arbitrary record-specific patterns.

`Customer Id` contains 20,652 unique values across 180,519 rows, indicating that customers have multiple records. Although customer identity might capture customer-specific patterns, using raw IDs could lead to memorization and poor generalization, particularly for unseen customers.

**Decision:** Both `Customer Id` and `Order Item Id` will be excluded from the initial model. If customer-level historical information becomes available at prediction time, meaningful aggregate features can be investigated separately.


```python
numerical_features = [
    'Days for shipment (scheduled)',
    'Order Item Discount',
    'Order Item Quantity',
    'Sales',
    'Product Price',
    'Latitude',
    'Longitude'
]

categorical_features = [
    'Type',
    'Customer Country',
    'Customer Segment',
    'Market',
    'Customer State',
    'Order Region',
    'Order Country',
    'Category Id',
    'Product Card Id',
    'Department Id'
]

date_features = [
    'Order Year',
    'Order Month',
    'Order Day',
    'Order Day of Week',
    'Order Hour'
]

target = 'Late_delivery_risk'


selected_features = (
    numerical_features
    + categorical_features
    + date_features
)

missing_features = [
    col for col in selected_features + [target]
    if col not in df.columns
]

print("Number of numerical features:", len(numerical_features))
print("Number of categorical features:", len(categorical_features))
print("Number of date features:", len(date_features))
print("Total selected features:", len(selected_features))
print("Missing columns:", missing_features)
```

<h3><b>Final machine learning feature matrix</b></h3>

```python
selected_features = (
    numerical_features
    + categorical_features
    + date_features
)

X = df[selected_features].copy()
y = df[target].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nX columns:")
for i, col in enumerate(X.columns, start=1):
    print(f"{i}. {col}")

print("\nTarget distribution:")
print(y.value_counts())
```

**Insight:** The final feature matrix `X` contains **180,519 observations and 22 selected features**, while the target vector `y` contains **180,519 observations**. Therefore, every observation has a corresponding target value and no observations were lost during feature selection. The target contains **98,977 late deliveries (54.83%)** and **81,542 non-late deliveries (45.17%)**, indicating a moderate class imbalance.

**Decision:** The 22 selected features will be used as the initial input feature set for model development. The target `Late_delivery_risk` will be used as the binary classification target.
