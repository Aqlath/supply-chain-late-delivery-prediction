## 2. Data Cleaning & Preparation

In this section, we clean and prepare the dataset for further analysis and machine learning by handling missing values, removing unusable and redundant columns, correcting data inconsistencies, converting data types, checking numerical anomalies, and removing potential data leakage.


```python
import numpy as np 
import pandas as pd
```

```python
data = pd.read_csv("D:/projects/supply chain project/datasets/raw/DataCoSupplyChainDataset.csv",encoding='latin1')

```

```python
data.head()
```

```python
data.info()
```

```python
#drop leakage columns and unusable columns
data = data.drop(columns=['Customer Password','Customer Email','Customer Fname',
                   'Customer Lname','Product Description','Product Image',
                   'Product Status','Order Id','Order Customer Id',
                   'Days for shipping (real)','Delivery Status','shipping date (DateOrders)',
                   'Order Status','Benefit per order','Order Item Profit Ratio',
                   'Order Profit Per Order'])
```

```python
data.shape
```

```python
#checking missing values
data.isnull().sum()
```

```python
#missing of order zipcode based on order country
missing_by_country = data.groupby('Order Country')['Order Zipcode'].apply(
    lambda x: x.isna().mean() * 100
)

missing_by_country.describe()
```

```python
#drop order zipcode column
data = data.drop(columns=['Order Zipcode'])
```

```python
data.shape
```

```python
#find the 3 missing customer zipcode column and understand how it is assosiate with other columns
data.loc[
    data['Customer Zipcode'].isna(),
    ['Customer City', 'Customer State', 'Customer Country', 'Customer Zipcode']
]
```

```python
#finding number of unique values in customer info columns
data[['Customer Zipcode',"Customer City","Customer State",'Customer Country']].nunique()
```

```python
#convert order date columns from object dtype to datetime dtype
data['order date (DateOrders)'] = pd.to_datetime(data['order date (DateOrders)'])
```

```python
#understanding unique values from differnet columns
print("Type column unique values : ", data['Type'].unique())
print("\n\ncustomer segment column unique values : ",data['Customer Segment'].unique())
print("\nMarket column unique values : ",data['Market'].unique())
print("\nShipping mode column unique values : ",data['Shipping Mode'].unique())
print("\nDepartment name column unique values : ",data['Department Name'].unique())
print("\nCategory Name column unique values : ", data['Category Name'].unique())
print("\nCustomer City column unique values : ", data['Customer City'].unique())
print("\nCustomer Country column unique values : ", data['Customer Country'].unique())
print("\nCustomer State column unique values : ", data['Customer State'].unique())
print("\nCustomer Street column unique values : ", data['Customer Street'].unique())
print("\nOrder City column unique values : ", data['Order City'].unique())
print("\nOrder Country column unique values : ", data['Order Country'].unique())
print("\nOrder Region column unique values : ", data['Order Region'].unique())
print("\nOrder State column unique values : ", data['Order State'].unique())
print("\nProduct Name column unique values : ", data['Product Name'].unique())
```

```python
#removing the leading and trailing spaces in the text from department name column
data['Department Name'] = (data['Department Name'].str.strip())
```

```python
data['Department Name'].unique()
```

```python
# Check for leading/trailing whitespace in all categorical columns

categorical_cols = data.select_dtypes(include='object').columns

for col in categorical_cols:
    whitespace_values = data[col].dropna().loc[
        data[col].dropna().str.strip() != data[col].dropna()
    ]
    
    if len(whitespace_values) > 0:
        print(f"\n{col}")
        print(whitespace_values.unique()[:20])
```

```python
#removing leading and trailiing spaces from the columns which contains spaces
categorical_cols = data.select_dtypes(include='object').columns

for col in categorical_cols:
    data[col] = data[col].str.strip()
```

```python
#find the customer state column which contain these values ('95758', '91732') and understand with other related columns
data[data['Customer State'].isin(['95758', '91732'])][
    ['Customer City', 'Customer State', 'Customer Country', 'Customer Zipcode']
]
```

```python
# Fix misplaced ZIP codes in Customer State

mask = data['Customer State'].isin(['95758', '91732'])

data.loc[mask, 'Customer Zipcode'] = data.loc[mask, 'Customer State']
data.loc[mask, 'Customer State'] = 'CA'
```

```python
#check missing values in each columns
missing = data.isnull().sum()

missing[missing > 0]
```

```python
#dropping redundant columns
data.drop(columns=[
    'Order Item Product Price',
    'Sales per customer',
    'Product Category Id',
    'Order Item Cardprod Id'
], inplace=True)
```

```python
#checking data type
data.dtypes
```

```python
#current columns
data.columns.tolist()
```

```python
categorical_cols = data.select_dtypes(include='object').columns

for col in categorical_cols:
    if col != 'Customer Zipcode':
        multiple_space_count = data[col].str.contains(
            r'\s{2,}',
            regex=True,
            na=False
        ).sum()

        if multiple_space_count > 0:
            print(f"{col}: {multiple_space_count} values with multiple spaces")
```

```python
categorical_cols = data.select_dtypes(include='object').columns

for col in categorical_cols:
    if col != 'Customer Zipcode':
        data[col] = (
            data[col]
            .str.replace(r'\s+', ' ', regex=True)
            .str.strip()
        )
```
